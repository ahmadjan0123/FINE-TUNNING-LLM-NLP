```python
from transformers import pipeline
```

<h1>Creating the model</h1>

```python
qa = pipeline('question-answering')
```

<H1>build context</h1>

```python
context = 'Pakistan got independence on 14 August,1947'
```

<h1>BUILD MODEL</h1>

```python
result = qa(
    question = 'WHEN DID PAKISATAN GOT INDEPENDENCE',
    context = context
)
```
