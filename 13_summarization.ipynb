```python
from transformers import pipeline 
```

<h1>SUMMARIZATION</h1>

```python
summarise = pipeline('summarization',model = 'facebook/bart-large-cnn')
```

```python
text  = '''Artificial Intelligence is transforming industries by automating tasks,
improving decision making, and enabling new applications.
AI is widely used in healthcare, finance, education,
transportation, and cybersecurity.'''
```

<h3>Generate result</h3>

```python
result = summarise(text, max_length = 30, min_length=10)
```

```python
print(result[0]['summary_text'])
```
