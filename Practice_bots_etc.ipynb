```python
from transformers import pipeline
```

```python
qa = pipeline('question-answering')
```

```python
context = ''' Pakistan got independence in 1947, pakistan is a country in south asia, imran khan is a pakistani hero'''
```

```python
result = qa(
    question = 'where is pakistan',
    context = context
)
```

```python
print(result)
```

```python
# predict next word
```

```python
next_word = pipeline('fill-mask',model ='bert-base-uncased' )
```

```python
word = next_word('the capital of UK is [MASK].')
```

```python
for r in word:
    print(r['token_str'],':',r['score'])
```

```python
# SUMMARIZATION
```

```python
summarise = pipeline('summarization')
```

```python
text = '''Artificial Intelligence is a broad field. it have taken every thing by fire. it is more of a bussiness transformation rather than a technological transformation. every one should learn it , adopt it and accept the change. the one that will not accpet the change will be left behind.'''
```

```python
ans = summarise(text, max_length = 30, min_length = 10)
```

```python
print(ans)
```
