<h1>IMPORTING LIBRARIES</h1>

```python
from transformers import pipeline
```

<h1>Creating a pipeline<h1>

```python
generator = pipeline(
    'text-generation',   # what di i want my model to do
    model = 'gpt2'       # which model do i want it to use
)
```

<h1>GENERATE TEXT</h1>

```python
result = generator(
    'Artificial intelligence is',
    max_new_tokens = 30,
    do_sample = True,
    temperature = 0.2
)
```

```python
print(result[0]['generated_text'])
```
